# Codigo

```python
import math

celsius = [0, 20, 37, 100]
lista1 = [1, 2, 3, 4, 5]
lista2 = [10, 20, 30, 40, 50]
radios = [1, 3, 5, 7, 10]
precios = {"pan": 1000, "leche": 2500, "café": 5000}
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

a = list(map(lambda c: (c * 9/5) + 32, celsius))


b = list(map(lambda x, y: x + y, lista1, lista2))


c = list(map(lambda r: math.pi * r**2, radios))

d = list(map(lambda item: (item[0], item[1] * 1.10), precios.items()))

e = list(map(lambda fila: list(map(lambda n: n * 10, fila)), matriz))


print(a)
print(b)
print(c)
print(d)
print(e)
```
