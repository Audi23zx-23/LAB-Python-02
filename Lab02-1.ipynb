# Codigo

```python
multiplo_3 = lambda x: x % 3 == 0
print(multiplo_3(9))  
print(multiplo_3(10)) 
```

```python
cubo = lambda x: x ** 3
print(cubo(2))
print(cubo(3))
```

```python
producto = lambda x, y: x * y
print(producto(4, 5))
```

```python
mayor = lambda x, y: x if x > y else y
print(mayor(10, 20))
```

```python
empieza_con_a = lambda s: s.lower().startswith('a')
print(empieza_con_a("Apple"))
print(empieza_con_a("Banana"))
```

```python
Celsius_a_fahrenheit = lambda c: (c * 9/5) + 32
print(Celsius_a_fahrenheit(0))
print(Celsius_a_fahrenheit(100))
```

```python
operaciones = [
    lambda n: n * 2,
    lambda n: n + 10,
    lambda n: n ** 2
]

numero = int(input("Ingresa un número: "))

for op in operaciones:
    print(op(numero))
```
