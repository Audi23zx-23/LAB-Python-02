```python
palabras = ["sol", "python", "casa", "computadora", "mar", "programacion"]
lista = [1, None, 3, None, 5, None, 7]
numeros = [10, 25, 33, 45, 62, 75, 80, 95]
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

a = list(filter(lambda p: len(p) > 4, palabras))

b = list(filter(lambda x: x is not None, lista))

c = list(filter(lambda p: p[0].lower() in "aeiou", palabras))

d = list(filter(lambda p: p == p[::-1], palabras))

e = list(filter(lambda n: n % 10 == 5, numeros))

f = list(filter(lambda p: p["precio"] > 100, productos))


print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```
