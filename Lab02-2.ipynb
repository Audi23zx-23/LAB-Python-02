# Codigo

```python
numeros = [10, 15, 3, 22, 8, 17, 6, 13]
palabras = ["sol", "python", "casa", "computadora", "mar", "programacion"]
celsius = [0, 20, 37, 100]
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

a = [(n, n**2) for n in numeros]

b = [n**3 for n in numeros if n > 10]

c = [n for n in numeros if n % 2 == 0]

d = [f"valor: {n}" for n in numeros]

e = [(c * 9/5) + 32 for c in celsius]

f = [p for p in palabras if len(p) > 5]

g = [p[0].upper() for p in palabras]

h = ["par" if n % 2 == 0 else "impar" for n in numeros]

i = [n for fila in matriz for n in fila]


print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
print(h)
print(i)
```
