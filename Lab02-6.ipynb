```python
def minimax(piedras, es_max):
    if piedras == 0:
        return 1 if es_max else -1  

    if es_max:
        mejor = -2
        for mov in [1, 2, 3]:
            if piedras - mov >= 0:
                valor = minimax(piedras - mov, False)
                mejor = max(mejor, valor)
        return mejor
    else:
        mejor = 2
        for mov in [1, 2, 3]:
            if piedras - mov >= 0:
                valor = minimax(piedras - mov, True)
                mejor = min(mejor, valor)
        return mejor


def movimiento_cpu(piedras):
    mejor_valor = -2
    mejor_mov = 1

    for mov in [1, 2, 3]:
        if piedras - mov >= 0:
            valor = minimax(piedras - mov, False)  
            if valor > mejor_valor:
                mejor_valor = valor
                mejor_mov = mov

    return mejor_mov


def jugar():
    piedras = 10
    print("=== JUEGO DE NIM ===")
    print("Quién no pueda mover, pierde.")
    print(f"Piedras iniciales: {piedras}\n")

    while True:
        print(f"Piedras restantes: {piedras}")
        while True:
            try:
                mov = int(input("Tu turno. ¿Cuántas piedras quitás? (1, 2 o 3): "))
                if mov in [1, 2, 3] and piedras - mov >= 0:
                    break
                else:
                    print("Movimiento inválido. Ingresá 1, 2 o 3 (y que haya suficientes piedras).")
            except ValueError:
                print("Ingresá un número válido.")

        piedras -= mov
        print(f"Quitaste {mov}. Piedras restantes: {piedras}\n")

        if piedras == 0:
            print("¡Ganaste! La CPU no puede mover.")
            break

        mov_cpu = movimiento_cpu(piedras)
        piedras -= mov_cpu
        print(f"La CPU quita {mov_cpu}. Piedras restantes: {piedras}\n")

        if piedras == 0:
            print("¡Perdiste! No podés mover.")
            break


jugar()
```
